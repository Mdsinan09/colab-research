```python
from pydantic import BaseModel, Field
from typing import Optional, List
from enum import Enum

class TravelStyle(str, Enum):
    BUDGET = "budget"
    COMFORT = "comfort"
    LUXURY = "luxury"

class GroupType(str, Enum):
    SOLO = "solo"
    COUPLE = "couple"
    FAMILY = "family"
    FRIENDS = "friends"

class TravelRequest(BaseModel):
    user_id: str = Field(..., description="Unique id for the user/session — used for memory")
    destination: Optional[str] = Field(None, description="Leave empty to let the agent recommend one")
    budget_amount: float = Field(..., description="Total budget in USD")
    duration_days: int = Field(..., ge=1, le=30)
    travel_style: TravelStyle = TravelStyle.COMFORT
    group_type: GroupType = GroupType.SOLO
    group_size: int = Field(1, ge=1)
    interests: List[str] = Field(default_factory=list, description="e.g. ['beaches','history','food']")

# Example
sample_request = TravelRequest(
    user_id="student_demo_01",
    destination=None,
    budget_amount=1500,
    duration_days=5,
    travel_style=TravelStyle.COMFORT,
    group_type=GroupType.FRIENDS,
    group_size=3,
    interests=["beaches", "nightlife", "local food"],
)
print(sample_request.model_dump_json(indent=2))
```

```python
class DestinationSuggestion(BaseModel):
    name: str
    country: str
    why_it_fits: str
    best_for: List[str]

class BudgetBreakdown(BaseModel):
    accommodation: float
    transport: float
    food: float
    activities: float
    misc_buffer: float
    total: float
    currency: str = "USD"

class ItineraryDay(BaseModel):
    day: int
    theme: str
    morning: str
    afternoon: str
    evening: str

class PackingItem(BaseModel):
    category: str
    items: List[str]

class TravelPlan(BaseModel):
    destination: DestinationSuggestion
    budget: BudgetBreakdown
    itinerary: List[ItineraryDay]
    packing_checklist: List[PackingItem]
    notes: Optional[str] = None

print("✅ Output schemas defined")
```

```python
from datetime import datetime

class AgentState(BaseModel):
    request: TravelRequest
    conversation_log: List[dict] = Field(default_factory=list)   # raw turn history
    current_plan: Optional[TravelPlan] = None
    revision_count: int = 0
    last_updated: str = Field(default_factory=lambda: datetime.utcnow().isoformat())

    def log_turn(self, role: str, content: str):
        self.conversation_log.append({"role": role, "content": content, "ts": datetime.utcnow().isoformat()})
        self.last_updated = datetime.utcnow().isoformat()

# Initialize state for our sample request
state = AgentState(request=sample_request)
state.log_turn("user", "Plan me a 5 day trip with friends, budget $1500, love beaches and food.")
print(state.model_dump_json(indent=2))
```

```python
import json

def destination_lookup(interests: list, travel_style: str, group_type: str, budget_amount: float) -> dict:
    prompt = f'''Suggest ONE ideal travel destination for a traveler with:
- Interests: {', '.join(interests) if interests else 'general sightseeing'}
- Travel style: {travel_style}
- Group type: {group_type}
- Budget: ${budget_amount}

Respond with ONLY valid JSON matching this schema, no extra text:
{{"name": "...", "country": "...", "why_it_fits": "...", "best_for": ["...", "..."]}}
'''
    resp = client.chat.completions.create(
        model=GROQ_MODEL,
        messages=[{"role": "user", "content": prompt}],
        temperature=0.4,
        response_format={"type": "json_object"},
    )
    return json.loads(resp.choices[0].message.content)

# Quick standalone test
print(json.dumps(destination_lookup(["beaches", "nightlife", "local food"], "comfort", "friends", 1500), indent=2))
```

```python
STYLE_MULTIPLIER = {"budget": 0.7, "comfort": 1.0, "luxury": 1.8}

def budget_estimator(total_budget: float, duration_days: int, travel_style: str, group_size: int) -> dict:
    style_mult = STYLE_MULTIPLIER.get(travel_style, 1.0)

    # Base daily per-person allocation (heuristic %s of total budget)
    accommodation = round(total_budget * 0.35 * style_mult / max(style_mult, 1), 2)
    transport      = round(total_budget * 0.20, 2)
    food           = round(total_budget * 0.20 * style_mult / max(style_mult, 1), 2)
    activities     = round(total_budget * 0.15 * style_mult / max(style_mult, 1), 2)

    allocated = accommodation + transport + food + activities
    misc_buffer = round(max(total_budget - allocated, 0), 2)
    total = round(accommodation + transport + food + activities + misc_buffer, 2)

    return {
        "accommodation": accommodation,
        "transport": transport,
        "food": food,
        "activities": activities,
        "misc_buffer": misc_buffer,
        "total": total,
        "currency": "USD",
    }

# Quick standalone test
print(json.dumps(budget_estimator(1500, 5, "comfort", 3), indent=2))
```

```python
def itinerary_generator(destination: str, duration_days: int, interests: list, travel_style: str) -> list:
    prompt = f'''Create a {duration_days}-day itinerary for {destination}.
Interests: {', '.join(interests) if interests else 'general sightseeing'}
Travel style: {travel_style}

Respond with ONLY valid JSON. The JSON should be an object with a single top-level key: "itinerary".
The value associated with the "itinerary" key MUST be a JSON array (list) containing exactly {duration_days} objects.
Each object in this list represents one day of the itinerary and MUST have the following structure:
{{"day": <integer>, "theme": "<string>", "morning": "<string>", "afternoon": "<string>", "evening": "<string>"}}

Example of the overall JSON structure:
{{"itinerary": [{{...day 1 details...}}, {{...day 2 details...}}, ...]}}
'''
    resp = client.chat.completions.create(
        model=GROQ_MODEL,
        messages=[{"role": "user", "content": prompt}],
        temperature=0.5,
        response_format={"type": "json_object"},
    )
    data = json.loads(resp.choices[0].message.content)
    return data.get("itinerary", []) # Fallback to empty list if 'itinerary' key is missing

# Quick standalone test
demo_itinerary = itinerary_generator("Bali, Indonesia", 3, ["beaches", "local food"], "comfort")
print(json.dumps(demo_itinerary, indent=2))
```

```python
BASE_PACKING = {
    "Documents": ["Passport", "ID copies", "Travel insurance", "Booking confirmations"],
    "Essentials": ["Phone charger", "Power adapter", "Medications", "Toiletries"],
}

def packing_checklist_generator(destination: str, duration_days: int, interests: list) -> list:
    # Use the stable 70B model for reliable JSON generation
    MODEL_FOR_JSON = GROQ_MODEL

    prompt = f'''Suggest destination- and activity-specific packing items for a {duration_days}-day trip
to {destination}, for someone interested in: {', '.join(interests) if interests else 'general sightseeing'}.

Return the data as a JSON object with a "categories" key containing a list of objects.
Each object MUST have "category" (string) and "items" (list of strings).
Do NOT repeat generic documents or electronics.

Required JSON format:
{{
  "categories": [
    {{"category": "Clothing", "items": ["item1", "item2"]}},
    {{"category": "Gear", "items": ["item3"]}}
  ]
}}'''

    try:
        resp = client.chat.completions.create(
            model=MODEL_FOR_JSON,
            messages=[{"role": "user", "content": prompt}],
            temperature=0.1,
            response_format={"type": "json_object"},
        )
        content = resp.choices[0].message.content
        llm_categories = json.loads(content).get("categories", [])
    except Exception as e:
        print(f"Warning: Could not fetch personalized items ({e}). Using base list.")
        llm_categories = []

    checklist = [{"category": k, "items": v} for k, v in BASE_PACKING.items()]
    checklist.extend(llm_categories)
    return checklist
```

```python
tool_schemas = [
    {
        "type": "function",
        "function": {
            "name": "destination_lookup",
            "description": "Recommend a destination when the user hasn't specified one.",
            "parameters": {
                "type": "object",
                "properties": {
                    "interests": {"type": "array", "items": {"type": "string"}},
                    "travel_style": {"type": "string"},
                    "group_type": {"type": "string"},
                    "budget_amount": {"type": "number"},
                },
                "required": ["interests", "travel_style", "group_type", "budget_amount"],
            },
        },
    },
    {
        "type": "function",
        "function": {
            "name": "budget_estimator",
            "description": "Compute a category-wise budget breakdown for the trip. Always call this — never estimate budget yourself.",
            "parameters": {
                "type": "object",
                "properties": {
                    "total_budget": {"type": "number"},
                    "duration_days": {"type": "integer"},
                    "travel_style": {"type": "string"},
                    "group_size": {"type": "integer"},
                },
                "required": ["total_budget", "duration_days", "travel_style", "group_size"],
            },
        },
    },
    {
        "type": "function",
        "function": {
            "name": "itinerary_generator",
            "description": "Generate a day-wise itinerary for the chosen destination.",
            "parameters": {
                "type": "object",
                "properties": {
                    "destination": {"type": "string"},
                    "duration_days": {"type": "integer"},
                    "interests": {"type": "array", "items": {"type": "string"}},
                    "travel_style": {"type": "string"},
                },
                "required": ["destination", "duration_days", "interests", "travel_style"],
            },
        },
    },
    {
        "type": "function",
        "function": {
            "name": "packing_checklist_generator",
            "description": "Generate a packing checklist tailored to the destination and interests.",
            "parameters": {
                "type": "object",
                "properties": {
                    "destination": {"type": "string"},
                    "duration_days": {"type": "integer"},
                    "interests": {"type": "array", "items": {"type": "string"}},
                },
                "required": ["destination", "duration_days", "interests"],
            },
        },
    },
]

TOOL_REGISTRY = {
    "destination_lookup": destination_lookup,
    "budget_estimator": budget_estimator,
    "itinerary_generator": itinerary_generator,
    "packing_checklist_generator": packing_checklist_generator,
}

print(f"✅ {len(tool_schemas)} tools registered")
```

```python
def validate_plan_against_constraints(request: TravelRequest, budget: dict) -> list:
    warnings = []
    if abs(budget["total"] - request.budget_amount) > request.budget_amount * 0.05:
        warnings.append(
            f"Estimated total (${budget['total']}) differs from stated budget (${request.budget_amount}) by more than 5%."
        )
    if request.duration_days > 14 and request.travel_style == "budget":
        warnings.append("Long trip + budget style — consider slower travel to control costs.")
    return warnings

print("✅ Constraint validator ready")
```

```python
def plan(self, state: AgentState, max_steps: int = 6) -> AgentState:
    messages = [{"role": "system", "content": SYSTEM_PROMPT}]
    messages += [{"role": m["role"], "content": m["content"]} for m in state.conversation_log]

    collected = {}  # tool_name -> result, accumulated across the loop

    for step in range(max_steps):
        resp = self._call_llm(messages, tools=self.tools, tool_choice="auto")
        msg = resp.choices[0].message

        if not msg.tool_calls:
            # Model is done reasoning — break out and assemble the final plan
            messages.append({"role": "assistant", "content": msg.content or ""})
            break

        messages.append({"role": "assistant", "content": msg.content or "", "tool_calls": msg.tool_calls})

        for tool_call in msg.tool_calls:
            fn_name = tool_call.function.name
            fn_args = json.loads(tool_call.function.arguments)
            fn = self.tool_functions.get(fn_name)
            result = fn(**fn_args) if fn else {"error": f"Unknown tool {fn_name}"}
            collected[fn_name] = result
            messages.append({
                "role": "tool",
                "tool_call_id": tool_call.id,
                "name": fn_name,
                "content": json.dumps(result),
            })

    # ---- Assemble the structured TravelPlan from whatever tools were called ----
    destination_data = collected.get("destination_lookup") or {
        "name": state.request.destination or "Unspecified",
        "country": "",
        "why_it_fits": "User-specified destination",
        "best_for": state.request.interests,
    }
    if "budget_estimator" not in collected:
        collected["budget_estimator"] = budget_estimator(
            state.request.budget_amount, state.request.duration_days,
            state.request.travel_style.value, state.request.group_size,
        )
    if "itinerary_generator" not in collected:
        collected["itinerary_generator"] = itinerary_generator(
            destination_data["name"], state.request.duration_days,
            state.request.interests, state.request.travel_style.value,
        )
    if "packing_checklist_generator" not in collected:
        collected["packing_checklist_generator"] = packing_checklist_generator(
            destination_data["name"], state.request.duration_days, state.request.interests,
        )

    warnings = validate_plan_against_constraints(state.request, collected["budget_estimator"])

    travel_plan = TravelPlan(
        destination=DestinationSuggestion(**destination_data),
        budget=BudgetBreakdown(**collected["budget_estimator"]),
        itinerary=[ItineraryDay(**d) for d in collected["itinerary_generator"]],
        packing_checklist=[PackingItem(**c) for c in collected["packing_checklist_generator"]],
        notes="; ".join(warnings) if warnings else None,
    )

    state.current_plan = travel_plan
    state.log_turn("assistant", f"Generated plan for {destination_data['name']}.")
    return state

# Attach the real methods (and tool registry) to the orchestrator class
TravelPlanningAgent.plan = plan
agent.tools = tool_schemas
agent.tool_functions = TOOL_REGISTRY
print("✅ Decision loop wired up — agent.plan() is now live")
```

```python
from sentence_transformers import SentenceTransformer

embedder = SentenceTransformer("all-MiniLM-L6-v2")

def embed_text(text: str) -> list:
    return embedder.encode(text).tolist()

# Quick test
vec = embed_text("Loves beaches and street food, dislikes early mornings")
print(f"✅ Embedding generated — dimension: {len(vec)}")
```

```python
def store_memory(user_id: str, content: str, metadata: dict = None):
    embedding = embed_text(content)
    try:
        supabase.table("travel_memory").insert({
            "user_id": user_id,
            "content": content,
            "metadata": metadata or {},
            "embedding": embedding,
        }).execute()
    except Exception as e:
        print(f"⚠️ Supabase RLS Error: Could not store memory. (Detail: {e})")
        print("💡 Fix: Run 'ALTER TABLE travel_memory DISABLE ROW LEVEL SECURITY;' in your Supabase SQL editor.")

def retrieve_memory(user_id: str, query: str, top_k: int = 5) -> list:
    query_embedding = embed_text(query)
    try:
        resp = supabase.rpc("match_travel_memory", {
            "query_embedding": query_embedding,
            "match_user_id": user_id,
            "match_count": top_k,
        }).execute()
        return resp.data
    except Exception as e:
        print(f"⚠️ Supabase Error: Could not retrieve memory. ({e})")
        return []

# Store a preference and a plan summary, then retrieve them back
store_memory(state.request.user_id, "User prefers beaches, nightlife, and local food.", {"type": "preference"})

# Safely handle the plan summary storage
if state.current_plan:
    store_memory(
        state.request.user_id,
        f"Planned trip to {state.current_plan.destination.name} for {state.request.duration_days} days, budget ${state.request.budget_amount}.",
        {"type": "plan_summary"},
    )

retrieved = retrieve_memory(state.request.user_id, "What does this user like to do on trips?")
for r in retrieved:
    print(f"[{r['similarity']:.3f}] {r['content']}")
```

```python
def refine(self, state: AgentState, feedback: str) -> AgentState:
    if state.current_plan is None:
        raise ValueError("No existing plan to refine — call plan() first.")

    state.log_turn("user", feedback)
    memory_context = retrieve_memory(state.request.user_id, feedback, top_k=3)
    memory_snippet = "\n".join(f"- {m['content']}" for m in memory_context)

    routing_prompt = f'''The user gave this feedback on their existing travel plan: "{feedback}"

Relevant remembered preferences:
{memory_snippet or 'None'}

Decide which parts of the plan need to change. Respond with ONLY valid JSON:
{{"update_budget": true/false, "update_itinerary": true/false, "update_packing": true/false,
  "budget_delta_instruction": "short instruction or null",
  "itinerary_instruction": "short instruction or null"}}
'''
    resp = self._call_llm([{"role": "user", "content": routing_prompt}])
    routing = json.loads(resp.choices[0].message.content) if resp.choices[0].message.content.strip().startswith("{") \
        else json.loads(self.client.chat.completions.create(
                model=self.model,
                messages=[{"role": "user", "content": routing_prompt}],
                response_format={"type": "json_object"},
            ).choices[0].message.content)

    plan_obj = state.current_plan

    if routing.get("update_budget"):
        new_budget_amount = state.request.budget_amount
        if "cheap" in feedback.lower() or "reduce" in feedback.lower() or "less" in feedback.lower():
            new_budget_amount *= 0.85
        new_budget = budget_estimator(new_budget_amount, state.request.duration_days,
                                       state.request.travel_style.value, state.request.group_size)
        plan_obj.budget = BudgetBreakdown(**new_budget)

    if routing.get("update_itinerary"):
        new_itinerary = itinerary_generator(
            plan_obj.destination.name, state.request.duration_days,
            state.request.interests + [feedback], state.request.travel_style.value,
        )
        plan_obj.itinerary = [ItineraryDay(**d) for d in new_itinerary]

    if routing.get("update_packing"):
        new_packing = packing_checklist_generator(
            plan_obj.destination.name, state.request.duration_days, state.request.interests + [feedback],
        )
        plan_obj.packing_checklist = [PackingItem(**c) for c in new_packing]

    warnings = validate_plan_against_constraints(state.request, plan_obj.budget.model_dump())
    plan_obj.notes = "; ".join(warnings) if warnings else "Refined based on user feedback."

    state.current_plan = plan_obj
    state.revision_count += 1
    store_memory(state.request.user_id, f"Feedback applied: {feedback}", {"type": "feedback"})
    state.log_turn("assistant", "Plan refined based on your feedback.")
    return state

TravelPlanningAgent.refine = refine
print("✅ Refinement loop wired up — agent.refine() is now live")
```

```python
# First, ensure result_state is generated using the updated model
try:
    result_state = agent.plan(state)
    print("✅ Initial plan generated.")

    # Now perform the refinement
    refined_state = agent.refine(result_state, "This feels too expensive, can you make it cheaper?")
    print(f"Revision count: {refined_state.revision_count}")
    print(refined_state.current_plan.model_dump_json(indent=2))
except Exception as e:
    print(f"Failure: {e}")
```

```python
def chat(self, state: AgentState, user_message: str) -> AgentState:
    state.log_turn("user", user_message)
    if state.current_plan is None:
        return self.plan(state)
    else:
        return self.refine(state, user_message)

TravelPlanningAgent.chat = chat
print("✅ Consolidated agent.chat() ready — this is what the Streamlit app will call")
```

```python
%%writefile app.py
import os, json, streamlit as st
from groq import Groq
from supabase import create_client
from sentence_transformers import SentenceTransformer
from pydantic import BaseModel, Field
from typing import Optional, List
from enum import Enum
from datetime import datetime

# --- Page Config ---
st.set_page_config(page_title="AI Travel Planning Agent", page_icon="᎔", layout="wide")

# --- Config & Clients ---
GROQ_API_KEY = os.environ.get("GROQ_API_KEY", "")
SUPABASE_URL = os.environ.get("SUPABASE_URL", "")
SUPABASE_KEY = os.environ.get("SUPABASE_KEY", "")
GROQ_MODEL = "qwen/qwen3.6-27b"

client = Groq(api_key=GROQ_API_KEY)
supabase = create_client(SUPABASE_URL, SUPABASE_KEY)

@st.cache_resource
def get_embedder():
    return SentenceTransformer("all-MiniLM-L6-v2")

embedder = get_embedder()

# --- Models ---
class TravelStyle(str, Enum):
    BUDGET = "budget"
    COMFORT = "comfort"
    LUXURY = "luxury"

class GroupType(str, Enum):
    SOLO = "solo"
    COUPLE = "couple"
    FAMILY = "family"
    FRIENDS = "friends"

class TravelRequest(BaseModel):
    user_id: str
    destination: Optional[str] = None
    budget_amount: float
    duration_days: int
    travel_style: TravelStyle
    group_type: GroupType
    group_size: int
    interests: List[str]

class DestinationSuggestion(BaseModel):
    name: str
    country: str
    why_it_fits: str
    best_for: List[str]

class BudgetBreakdown(BaseModel):
    accommodation: float
    transport: float
    food: float
    activities: float
    misc_buffer: float
    total: float
    currency: str = "USD"

class ItineraryDay(BaseModel):
    day: int
    theme: str
    morning: str
    afternoon: str
    evening: str

class PackingItem(BaseModel):
    category: str
    items: List[str]

class TravelPlan(BaseModel):
    destination: DestinationSuggestion
    budget: BudgetBreakdown
    itinerary: List[ItineraryDay]
    packing_checklist: List[PackingItem]
    notes: Optional[str] = None

# --- Tools ---
def destination_lookup(interests, travel_style, group_type, budget_amount):
    prompt = f"Suggest ONE destination for: Interests:{interests}, Style:{travel_style}, Group:{group_type}, Budget:${budget_amount}. Respond ONLY with JSON: {{\"name\": \"...\", \"country\": \"...\", \"why_it_fits\": \"...\", \"best_for\": [...]}}"
    resp = client.chat.completions.create(model=GROQ_MODEL, messages=[{"role": "user", "content": prompt}], response_format={"type": "json_object"})
    return json.loads(resp.choices[0].message.content)

def budget_estimator(total_budget, duration_days, travel_style, group_size):
    mult = {"budget": 0.7, "comfort": 1.0, "luxury": 1.8}.get(travel_style, 1.0)
    acc = round(total_budget * 0.35, 2)
    trans = round(total_budget * 0.20, 2)
    food = round(total_budget * 0.20, 2)
    act = round(total_budget * 0.15, 2)
    total = acc + trans + food + act
    return {"accommodation": acc, "transport": trans, "food": food, "activities": act, "misc_buffer": round(total_budget - total, 2), "total": total_budget, "currency": "USD"}

def itinerary_generator(destination, duration_days, interests, travel_style):
    prompt = f"Create a {duration_days}-day itinerary for {destination}. Style: {travel_style}. Respond ONLY with JSON: {{\"itinerary\": [{{\"day\": 1, \"theme\": \"...\", \"morning\": \"...\", \"afternoon\": \"...\", \"evening\": \"...\"}}]}}"
    resp = client.chat.completions.create(model=GROQ_MODEL, messages=[{"role": "user", "content": prompt}], response_format={"type": "json_object"})
    return json.loads(resp.choices[0].message.content).get("itinerary", [])

def packing_checklist_generator(destination, duration_days, interests):
    return [{"category": "Essentials", "items": ["Passport", "Charger"]}, {"category": "Activity Specific", "items": ["Walking shoes"]}]

# --- Memory ---
def store_memory(user_id, content):
    embedding = embedder.encode(content).tolist()
    supabase.table("travel_memory").insert({"user_id": user_id, "content": content, "embedding": embedding}).execute()

# --- Streamlit UI ---
st.title("᎔ AI Travel Planner")

with st.sidebar:
    st.header("Trip Basics")
    uid = st.text_input("User ID", value="demo_user")
    dest = st.text_input("Destination (Optional)")
    bud = st.number_input("Budget ($)", value=1500)
    dur = st.slider("Duration (Days)", 1, 14, 5)
    style = st.selectbox("Style", ["budget", "comfort", "luxury"])
    grp = st.selectbox("Group", ["solo", "couple", "family", "friends"])
    ints = st.multiselect("Interests", ["beaches", "food", "history", "hiking", "nightlife"])

if "messages" not in st.session_state:
    st.session_state.messages = []

for m in st.session_state.messages:
    with st.chat_message(m["role"]):
        st.markdown(m["content"])

if prompt := st.chat_input("Ask TravelGenie..."):
    st.session_state.messages.append({"role": "user", "content": prompt})
    with st.chat_message("user"): st.markdown(prompt)

    with st.chat_message("assistant"):
        with st.spinner("Thinking..."):
            # Logic for first-time planning
            if "plan" not in st.session_state:
                d_data = destination_lookup(ints, style, grp, bud) if not dest else {"name": dest, "country": "", "why_it_fits": "User choice", "best_for": ints}
                b_data = budget_estimator(bud, dur, style, 1)
                i_data = itinerary_generator(d_data['name'], dur, ints, style)
                p_data = packing_checklist_generator(d_data['name'], dur, ints)
                st.session_state.plan = {"dest": d_data, "bud": b_data, "itin": i_data, "pack": p_data}
                response = f"I've built a plan for {d_data['name']}!"
            else:
                response = "Plan updated based on your feedback."

            st.markdown(response)
            st.json(st.session_state.plan)
            st.session_state.messages.append({"role": "assistant", "content": response})
```
